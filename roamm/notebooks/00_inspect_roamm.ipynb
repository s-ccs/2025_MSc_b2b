# To check whether to use right eye or left eye


```python
from pathlib import Path
import pandas as pd

files = sorted(
    Path("/scratch/data/ROAMM/derivatives/synced")
    .glob("sub-*/*_mldata.pkl")
)

rows = []

for file in files:
    df = pd.read_pickle(file)

    subject = file.parent.name
    run = int(df["run_num"].dropna().iloc[0])

    for eye in ["L", "R"]:
        tstart = f"fix_{eye}_tStart"
        word_key = f"fix_{eye}_fixed_word_key"

        if tstart not in df.columns or word_key not in df.columns:
            print(f"Skip {subject} run {run}, eye {eye}: columns missing")
            continue

        first_pass = (
            df.loc[
                df["first_pass_reading"].eq(True)
                & df[tstart].notna(),
                [tstart, word_key]
            ]
            .drop_duplicates(tstart)
        )

        rows.append({
            "subject": subject,
            "run": run,
            "eye": eye,
            "first_pass_fixations": len(first_pass),
            "valid_lexical_fixations": first_pass[word_key].notna().sum(),
            "missing_word": first_pass[word_key].isna().sum(),
        })

summary = pd.DataFrame(rows)
```
